```python
import numpy as np
import pandas as pd

import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(5,3.8),facecolor='white')
ax = plt.subplot(111)

p = 1e-6

D = np.logspace(4,10,150)
N = np.array([1e6,1e7,3e7,5e7,7e7,9e7,1e8])

plt.plot(D,np.sqrt(D*p/11), color='black', linewidth=.2)
plt.scatter(10*N, np.sqrt(10*p*N/11), color='black')

for n in N:
    plt.plot(x, np.sqrt(p*x/(1+x/n)))

        
plt.xscale('log')
#plt.yscale('log')
plt.xlim(1e4,1e10)
plt.ylim(0,1e4*np.sqrt(p))
plt.xticks(fontsize=15)
plt.yticks(fontsize=15)
ax.spines[['right', 'top']].set_visible(False)
plt.tight_layout()
plt.savefig('SNR_vs_depth_all.png',dpi=1200)
```

```python
plt.figure(figsize=(5,3.8),facecolor='white')
ax = plt.subplot(111)

p = 1e-6

D = np.logspace(4,10,150)
N = np.array([1e6,1e7,3e7,5e7,7e7,9e7,1e8])

plt.plot(D,np.sqrt(D*p/11), color='black', linewidth=.2,alpha=0.2)
plt.scatter(10*N, np.sqrt(10*p*N/11), color='black', alpha=0.2)

for n in N[:-1]:
    plt.plot(x, np.sqrt(p*x/(1+x/n)), alpha=0.2)

n = N[-1]
plt.plot(x, np.sqrt(p*x/(1+x/n)), alpha=0.2)
        
plt.xscale('log')
#plt.yscale('log')
plt.xlim(1e4,1e10)
plt.ylim(0,1e4*np.sqrt(p))
plt.xticks(fontsize=15)
plt.yticks(fontsize=15)
ax.spines[['right', 'top']].set_visible(False)
plt.tight_layout()
plt.savefig('SNR_vs_depth_one.png',dpi=1200)
```
