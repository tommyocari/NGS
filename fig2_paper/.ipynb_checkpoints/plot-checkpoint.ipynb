```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

#alpha = N/k

N_0 = 1539351
F = 8.922473368689328
```

```python
def a_n_function(alpha): # a_n-1 = epsilon
    n = 0
    S = 1
    alphas = [alpha]
    probs = []
    
    for i in range(100):
        p = 1./(1+alpha)
        probs.append(p) 
        q = 1-p
        s = 1+p
        S = S*s
        f = p*q/s
        n = n*s + f
        alpha = alpha*s; alphas.append(alpha)
        epsilon = n/S
    
    return epsilon, alphas, probs
```

```python
def N_function(N_0, p=0.6):
    return N_0*(1+p)**np.arange(100)
```

# Flat p

## number of DNA vs rounds

```python
plt.figure(figsize=(4,3),facecolor='white')


N_vector = np.array(N_function(N_0=N_0,p=0.6))

plt.plot(N_vector)

plt.yscale('log')
plt.xlim(0,50)
plt.xlabel('amplification round')
plt.ylabel('number of DNA molecules')
plt.tight_layout()
plt.savefig('pconst_N_vs_amp_rounds.png')
```

## probability vs rounds

## variance vs mean

```python
df_bins = pd.read_csv('../simulations_p_constant/df_bins_simulation_Byrne_25amp_rounds.csv',index_col=0)

plt.figure(figsize=(4,4), facecolor='white')
x = df_bins['mean_count']
y = df_bins['var_count']
plt.plot(x,x, linestyle='--', linewidth=1)
plt.plot(x, F*x,linestyle='--', color='black', linewidth=1)
plt.scatter(x,y, marker='.')
plt.xscale('log')
plt.yscale('log')
plt.xlabel('mean count')
plt.ylabel('variance count')
plt.tight_layout()
plt.savefig('pconst_var_vs_mean.png')
```

```python
x = np.arange(0,30,5)
y = np.zeros(6)
y_err = np.zeros(6)
i = 0
for amp_rounds in x:
    df_bins = pd.read_csv('../simulations_p_constant/df_bins_simulation_Byrne_%damp_rounds.csv'%amp_rounds)
    F_vec = ((df_bins['var_count'] - df_bins['var_lambda']) / df_bins['mean_count']).mean()
    y[i] = F_vec.mean()
    y_err[i] = F_vec.var()
    i+=1
    
plt.figure(figsize=(4,3), facecolor='white')
plt.errorbar(x,y,yerr=y_err, marker='o',color='C4')
plt.hlines(1,x[0],x[-1],linestyle='--')
plt.hlines(F,x[0],x[-1],linestyle='--', color='black')
plt.xlabel('amplification round')
plt.ylabel('noise extent')
plt.tight_layout()
plt.savefig('pconst_F_vs_amp_rounds.png')
```

# Michaelis Menten

## number of DNA vs rounds

```python
plt.figure(figsize=(4,3),facecolor='white')

for alpha in np.array([1e-5,1e-7,1e-9]):
    N_vector = N_0/alpha*np.array(a_n_function(alpha)[1])

    plt.plot(N_vector, label=' k = %.1e'%(N_0/alpha))

plt.legend()
plt.yscale('log')
plt.xlim(0,50)
plt.xlabel('amplification round')
plt.ylabel('number of DNA molecules')
plt.tight_layout()
plt.savefig('pMM_N_vs_amp_rounds.png')
```

## probability vs rounds

```python
plt.figure(figsize=(4,3),facecolor='white')

for alpha in np.array([1e-5,1e-7,1e-9]):
    p_vector = np.array(a_n_function(alpha)[2])

    plt.plot(p_vector, label=' k = %.1e'%(N_0/alpha))

plt.legend()
plt.xlim(0,50)
plt.xlabel('amplification round')
plt.ylabel('duplication probability')
plt.tight_layout()
plt.savefig('pMM_p_vs_amp_rounds.png')
```
