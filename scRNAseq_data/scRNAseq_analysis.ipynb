```python
import numpy as np
import pandas as pd
```

```python
datasets = ['HBreastCancer','HLungCancer','HPBMC2','PBMCs','MBrain','MBrainCoronal']
#datasets = ['HBreastCancer','HPBMC2','MBrain','MBrainCoronal']
df = pd.DataFrame(index = datasets, columns = ['mean','variance','F','R','N_0','sat_frac','uniques'])
for dataset in datasets:
    data = pd.read_csv('/home/tommaso/new_phd/scRNAseq/data/'+dataset+'/UR_counts.csv', names=['seq','count'])
    uniques = len(data.query('count > 0'))
    mean = data['count'].sum() / (4.**12)
    variance = (((data['count'] - mean)**2).sum() + (4**12 - len(data))*(mean**2))/(4.**12)
    F = variance / mean
    R = data['count'].sum()
    N_0 = R / (F-1)
    sat_frac = np.sqrt((F-1)/F)
    df.loc[dataset] = [mean, variance, F, R, N_0, sat_frac,uniques]
    print(dataset)
```

```python
F = np.loadtxt('../binning_real_data_q/F_Dalkara.csv')[0]
data = pd.read_csv('/home/tommaso/new_phd/data/Dalkara.csv',index_col=0)
R = data['T0'].sum()
uniques = len(data.query('T0 > 0'))
N_0 = R / (F-1)
sat_frac = np.sqrt((F-1)/F)
df.loc['DalkaraT0'] = [np.nan, np.nan, F, R, N_0, sat_frac,uniques]
```
