```python
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
```

```python
df_bins_original = pd.read_csv('../binning_real_data_q/df_bins_Dalkara_IM_200bins.csv',index_col=0)
df_bins_4t = pd.read_csv('../binning_real_data_q/df_bins_Dalkara_IM_2t.csv',index_col=0)
df_bins_16t = pd.read_csv('../binning_real_data_q/df_bins_Dalkara_IM_4t.csv',index_col=0)

plt.figure(figsize=(10,4), facecolor='white')

ax = plt.subplot(1,2,1)

df_bins = df_bins_original
x = df_bins['mean_lambda']
y = (df_bins['var_count'] - df_bins['var_lambda']) / df_bins['mean_lambda'] - 1
plt.scatter(x,y,marker='.')

df_bins = df_bins_4t
x = df_bins['mean_lambda']
y = (df_bins['var_count'] - df_bins['var_lambda']) / df_bins['mean_lambda'] - 1
plt.scatter(x,y,marker='.')

df_bins = df_bins_16t
x = df_bins['mean_lambda']
y = (df_bins['var_count'] - df_bins['var_lambda']) / df_bins['mean_lambda'] - 1
plt.scatter(x,y,marker='.')

plt.xscale('log')
plt.yscale('log')
plt.xlabel('mean lambda')
plt.ylabel('Fano factor - 1')

a = 5e-3
b = 1e1
c = 1e-1
d = 1e1

ax.add_patch(mpl.patches.Rectangle((a,c), width=b, height=d, angle=0.0, fill=False))

plt.subplot(1,2,2)

df_bins = df_bins_original
x = df_bins['mean_lambda']
y = (df_bins['var_count'] - df_bins['var_lambda']) / df_bins['mean_lambda'] - 1
plt.scatter(x,y,marker='.')

df_bins = df_bins_4t
x = df_bins['mean_lambda']
y = (df_bins['var_count'] - df_bins['var_lambda']) / df_bins['mean_lambda'] - 1
plt.scatter(x,y,marker='.')

df_bins = df_bins_16t
x = df_bins['mean_lambda']
y = (df_bins['var_count'] - df_bins['var_lambda']) / df_bins['mean_lambda'] - 1
plt.scatter(x,y,marker='.')

plt.xscale('log')
plt.yscale('log')
plt.xlabel('mean lambda')
plt.ylabel('Fano factor - 1')
plt.xlim(a,b)
plt.ylim(c,d)

plt.tight_layout()
plt.savefig('Dalkara_different_depth.png')
```
