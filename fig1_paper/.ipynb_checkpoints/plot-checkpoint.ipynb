```python
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score
```

## binning_IM

```python
bins=200
df_bins = pd.read_csv('../binning_real_data_q/df_bins_Dalkara_IM_%dbins.csv'%bins, index_col=0)

plt.figure(figsize=(8,8), facecolor='white')

plt.subplot(2,2,1)
x = df_bins['mean_lambda']
y = df_bins['mean_count']
plt.scatter(x, y, marker='.')
plt.plot(x, x, linestyle='--', color='red')
plt.xscale('log')
plt.yscale('log')
plt.xlabel('mean lambda')
plt.ylabel('mean count')

plt.subplot(2,2,2)
x = df_bins['mean_count']
y = df_bins['var_count']
plt.scatter(x, y, marker='.', color='red')
plt.plot(x, x, linestyle='--', color='black')
plt.xscale('log')
plt.yscale('log')
plt.xlabel('mean count')
plt.ylabel('var count')

a = 1e-1
b = 1e3
c = 1
d = 2e1

ax = plt.subplot(2,2,3)
x = df_bins['mean_count']
y1 = df_bins['var_count']
y2 = df_bins['var_lambda']
y1 = y1 / x
y2 = y2 / x
y3 = y1 - y2
plt.scatter(x, y1, marker='.', alpha=0.5)
plt.scatter(x, y2, marker='.', alpha=0.5)
plt.scatter(x, y3, marker='.', alpha=0.5)
plt.xscale('log')
plt.yscale('log')
plt.xlabel('mean lambda')
plt.ylabel('Fano factor')
plt.hlines(1, x.iloc[0], x.iloc[-1], color='black', linestyle='--')
ax.add_patch(mpl.patches.Rectangle((a,c), width=b, height=d, angle=0.0, fill=False))

plt.subplot(2,2,4)
x = df_bins['mean_count']
y1 = df_bins['var_count']
y2 = df_bins['var_lambda']
y1 = y1 / x
y2 = y2 / x
y3 = y1 - y2
plt.scatter(x, y1, marker='.', alpha=0.5)
plt.scatter(x, y2, marker='.', alpha=0.5)
plt.scatter(x, y3, marker='.', alpha=0.5)
plt.xscale('log')
plt.yscale('log')
plt.xlabel('mean count')
plt.ylabel('Fano factor')
plt.xlim(a,b)
plt.ylim(c,d)
plt.hlines(1, x.iloc[0], x.iloc[-1], color='black', linestyle='--')

plt.tight_layout()
plt.savefig('binning_IM_Dalkara_%dbins.png'%bins)
```

## Comparison IM and Potts

```python
df_bins_IM = pd.read_csv('../binning_real_data_q/df_bins_Dalkara_IM_200bins.csv', index_col=0)
df_bins_Potts = pd.read_csv('../binning_real_data_q/df_bins_Dalkara_Potts.csv', index_col=0)

plt.figure(figsize=(10,15))

plt.subplot(4,2,1)
x = df_bins_IM['mean_lambda']
y = df_bins_IM['mean_count']
R2 = r2_score(np.log(x),np.log(y))
plt.scatter(x, y, marker='.')
plt.plot(x, x, linestyle='--', color='red')
plt.xscale('log')
plt.yscale('log')
plt.xlabel('mean lambda')
plt.ylabel('mean count')
plt.title('Indepedent model\nR2 = %.6f'%R2)

plt.subplot(4,2,2)
x = df_bins_Potts['mean_lambda']
y = df_bins_Potts['mean_count']
R2 = r2_score(np.log(x),np.log(y))
plt.scatter(x, y, marker='.')
plt.plot(x, x, linestyle='--', color='red')
plt.xscale('log')
plt.yscale('log')
plt.xlabel('mean lambda')
plt.ylabel('mean count')
plt.title('Potts model\nR2 = %.6f'%R2)

a = 1e0
b = 5e2
c = 1
d = 1e2

ax = plt.subplot(4,2,3)
x = df_bins_IM['mean_count']
y1 = df_bins_IM['var_count']
y2 = df_bins_IM['var_lambda']
y1 = y1 / x
y2 = y2 / x
y3 = y1 - y2
plt.scatter(x, y1, marker='.', alpha=0.5)
plt.scatter(x, y2, marker='.', alpha=0.5)
plt.scatter(x, y3, marker='.', alpha=0.5)
plt.xscale('log')
plt.yscale('log')
plt.xlabel('mean count')
plt.ylabel('Fano factor')
plt.title('Indepedent model')
plt.hlines(1, x.iloc[0], x.iloc[-1], color='black', linestyle='--')
ax.add_patch(mpl.patches.Rectangle((a,c), width=b, height=d, angle=0.0, fill=False))


plt.subplot(4,2,5)
x = df_bins_IM['mean_count']
y1 = df_bins_IM['var_count']
y2 = df_bins_IM['var_lambda']
y1 = y1 / x
y2 = y2 / x
y3 = y1 - y2
plt.scatter(x, y1, marker='.', alpha=0.5)
plt.scatter(x, y2, marker='.', alpha=0.5)
plt.scatter(x, y3, marker='.', alpha=0.5)
plt.xscale('log')
plt.yscale('log')
plt.xlabel('mean lambda')
plt.ylabel('Fano factor')
plt.title('Indepedent model')
plt.xlim(a,b)
plt.ylim(c,d)
plt.hlines(1, x.iloc[0], x.iloc[-1], color='black', linestyle='--')


ax = plt.subplot(4,2,4)
x = df_bins_Potts['mean_count']
y1 = df_bins_Potts['var_count']
y2 = df_bins_Potts['var_lambda']
y1 = y1 / x
y2 = y2 / x
y3 = y1 - y2
plt.scatter(x, y1, marker='.', alpha=0.5)
plt.scatter(x, y2, marker='.', alpha=0.5)
plt.scatter(x, y3, marker='.', alpha=0.5)
plt.xscale('log')
plt.yscale('log')
plt.xlabel('mean count')
plt.ylabel('Fano factor')
plt.title('Potts model')
plt.hlines(1, x.iloc[0], x.iloc[-1], color='black', linestyle='--')
ax.add_patch(mpl.patches.Rectangle((a,c), width=b, height=d, angle=0.0, fill=False))

plt.subplot(4,2,6)
x = df_bins_Potts['mean_count']
y1 = df_bins_Potts['var_count']
y2 = df_bins_Potts['var_lambda']
y1 = y1 / x
y2 = y2 / x
y3 = y1 - y2
plt.scatter(x, y1, marker='.', alpha=0.5)
plt.scatter(x, y2, marker='.', alpha=0.5)
plt.scatter(x, y3, marker='.', alpha=0.5)
plt.xscale('log')
plt.yscale('log')
plt.xlabel('mean count')
plt.ylabel('Fano factor')
plt.title('Potts model')
plt.xlim(a,b)
plt.ylim(c,d)
plt.hlines(1, x.iloc[0], x.iloc[-1], color='black', linestyle='--')

plt.tight_layout()
plt.savefig('comparison_IM_Potts_Dalkara.png')
```
