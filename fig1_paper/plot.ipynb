```python
import numpy as np
import pandas as pd
import matplotlib as mpl
import matplotlib.pyplot as plt
from sklearn.metrics import r2_score
```

```python
bins=200
df_bins = pd.read_csv('../binning_real_data_q/df_bins_Dalkara_IM_%dbins.csv'%bins, index_col=0)

a,b = 1e-4,1e3

plt.figure(figsize=(4,3.8),facecolor='white')
ax = plt.subplot(111)
x = df_bins['mean_lambda']
y = df_bins['mean_count']
plt.scatter(x,y,marker='o')
plt.plot([a,b],[a,b],linestyle='--', color='red')
plt.xscale('log')
plt.yscale('log')
plt.yticks([1e-4, 1e-2, 1e0, 1e2],fontsize=15); plt.minorticks_off()
plt.xticks(fontsize=15)
plt.xlim(1e-4,1e3)
plt.ylim(1e-4,1e3)
ax.spines[['right', 'top']].set_visible(False)
plt.tight_layout()
plt.savefig('Dalkara_count_vs_lambda.png', dpi=1200)
```
