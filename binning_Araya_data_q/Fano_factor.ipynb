```python
import numpy as np
import pandas as pd
```

```python
df_F = pd.DataFrame(index=['T0_rep1','T0_rep2'],columns=['F'])
```

```python
replicate = 'T0_rep2'

F_array = []

for ham_dist in [2,3,4,5,6,7]:
    df_bins = pd.read_csv('df_bins_Araya_%s_ham%d.csv'%(replicate, ham_dist),index_col=0)
    for i in df_bins['F']:
        F_array.append(i)

F_array = np.array(F_array)
df_F.loc[replicate,'F'] = F_array.mean() 
```

```python
df_F.to_csv('F_Araya.csv')
```
